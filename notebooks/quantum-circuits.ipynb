# Quantum circuits

## Logic gates

- **Pauli-X gate**:
bit flip gate
$$X = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}$$

- **Pauli-Y gate**:
combination of bit flip and phase flip with an imaginary phase factor
$$Y = \begin{bmatrix} 0 & -i \\ i & 0 \end{bmatrix}$$

- **Pauli-Z gate**:
phase flip gate
$$Z = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$$
