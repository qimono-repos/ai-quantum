```python
import numpy as np
from qiskit.primitives import StatevectorSampler
from qiskit_machine_learning.kernels import FidelityQuantumKernel
from qiskit_machine_learning.state_fidelities import ComputeUncompute
from qiskit.circuit.library import ZZFeatureMap
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import make_moons
```

```python

# ==========================
# Setup
# ==========================
sampler = StatevectorSampler()                     # exact statevector simulation
feature_map = ZZFeatureMap(feature_dimension=2, reps=2)

# Official fidelity calculator (this is what your custom function was doing)
fidelity = ComputeUncompute(sampler=sampler)

# Quantum kernel
kernel = FidelityQuantumKernel(
    feature_map=feature_map,
    fidelity=fidelity
)

# ==========================
# Data (moons dataset)
# ==========================
X, y = make_moons(n_samples=100, noise=0.3, random_state=42)
X = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# ==========================
# QSVM
# ==========================
qsvc = SVC(kernel=kernel.evaluate)   # .evaluate turns the kernel into a callable
qsvc.fit(X_train, y_train)

print("Training accuracy:", qsvc.score(X_train, y_train))
print("Test accuracy:    ", qsvc.score(X_test, y_test))
```
