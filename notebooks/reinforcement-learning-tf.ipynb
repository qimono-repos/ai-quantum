# Reinforcement Learning with TensorFlow

The library `gym` is referneced in some courses. For modernizarion, to use libraries used currently, we import gymnasium

```python
import numpy as np
import gymnasium
```

```python
env = gymnasium.make("FrozenLake-v1", is_slippery="False")
state_size = env.observation_space.n
action_size = env.action_space.n
 
```

```python
Q_table = np.zeros((state_size, action_size))
print(Q_table)
```

alpha : learning rate; gamma : discount factor; epsilon : exploration rate; episodes: number of games

```python
alpha = 0.1
gamma = 0.99
epsilon = 1.0
espsilon_decay = 0.995
epsilon_min = 0.01
episodes = 2000
```

```python
for episode in range(episodes):
    state = env.reset()[0]
    done = False

    while not done:
        if np.random.rand() < epsilon:
            action = np.random.choice(action_size) #Explore
        else:
            action = np.argmax(Q_table[state]) #Exploit

        next_state, reward, done, _, _ = env.step(action)

        Q_table[state, action] = Q_table[state, action] + alpha * (reward + gamma * np.max(Q_table[next_state]) - Q_table[state, action])
        state = next_state
    
    epsilon = max(epsilon * espsilon_decay, epsilon_min)

print("Trained Q-table:")

    
```
