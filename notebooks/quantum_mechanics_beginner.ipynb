## wave function
The wave function, denoted as ψ (psi), is a fundamental concept in quantum mechanics.

### Probability and Normalization
Formula written with LaTeX math in Jupyter Markdown cells. Inline: $P(x)=|\psi(x)|^2$.
Display (centered):
$$P(x)=\left|\psi(x)\right|^2$$
And the normalization (integral) is:
$$\int |\psi(x)|^2 \, dx = 1$$

```python
# Programmatic rendering using IPython.display
from IPython.display import display, Math
display(Math(r'P(x)=|\psi(x)|^2'))
display(Math(r'\int |\psi(x)|^2 \, dx = 1'))
```

A qbit exists in a superposition of states.

$$|\psi\rangle = \alpha|0\rangle + \beta|1\rangle$$

$$|\alpha|^2 + |\beta|^2 = 1$$

A general qbit state is 

$$|\psi\rangle = \cos\left(\frac{\theta}{2}\right) \ket{0} + e^{i\phi} \sin\left(\frac{\theta}{2}\right) \ket{1}$$

**Bloch-sphere coordinates**:
The Cartesian coordinates of the Bloch vector for state $|\psi\rangle$ are:
$$ (x,y,z) = (\sin\theta\cos\phi,\; \sin\theta\sin\phi,\; \cos\theta) $$
You can compute these from `theta` and `phi` (radians) in a code cell below.

```python
import math
def bloch_coords(theta, phi):
    x = math.sin(theta) * math.cos(phi)
    y = math.sin(theta) * math.sin(phi)
    z = math.cos(theta)
    return x, y, z
# Example: change theta/phi as needed (radians)
theta = math.pi/3  # 60 degrees
phi = math.pi/4    # 45 degrees
x, y, z = bloch_coords(theta, phi)
print(f'theta={theta:.3f}, phi={phi:.3f}')
print(f'Bloch coords: x={x:.3f}, y={y:.3f}, z={z:.3f}')
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

- Create a quantum circuit with one qbit and one classical bit
- Apply a Hadamard gate to the qbit to create a superposition state
- Measure the qbit and store the result in the classical bit

```python
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)
```

```python

simulator = Aer.get_backend('aer_simulator')
job = simulator.run(qc, shots=1024)
result = job.result()
```

```python
from IPython.display import display
counts = result.get_counts(qc)

print(counts)
fig = plot_histogram(counts)
display(fig)
#plt.show()
```
